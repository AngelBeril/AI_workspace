```python
#print odd numbers in the list
```

```python
list3 = [20,10,16,19,25,1,276,188]
for tempList in list3:
    if(tempList % 2 != 0):
        print(f"{tempList} is odd")
```

```python
#print even numbers in the list
```

```python
for tempList in list3:
    if(tempList % 2 == 0):
        print(f"{tempList} is even")
```
